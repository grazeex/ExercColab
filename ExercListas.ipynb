1. Crie um programa que receba uma lista de números e retorne o maior valor dessa lista

## Algoritimos
1. Cria uma lista de números
2. Percorrer a lista em um loop para salvar o maior valor em uma variável
3.


```python
lista = [3, 6, 5, 9, 2, 1]
maior_valor = 0
for numero in lista:
  if numero > maior_valor:
    maior_valor = numero
print(maior_valor)
```
